```python
#inspecting unique values in all columns except 'Transaction Id'
df_cols = set(df.columns)
exclude = {'Transaction ID'}
df_cols = df_cols.difference(exclude)
df_cols = list(df_cols)

for col in df_cols:
  print(f'{col}:\n{df[col].unique()}', end = '\n\n')
```

```python
#replacing error and unknown
df.replace(['ERROR', 'UNKNOWN'], np.nan)
```

```python
df = df.replace(['ERROR', 'UNKNOWN'], np.nan)

df.head(10)
```

```python
#changing quantity as integer as it contains nan
df['Quantity'].astype('Int64')
```

```python
#changed all the below data types for accuarcy

df['Quantity'] = df['Quantity'].astype('Int64')
df[['Price Per Unit', 'Total Spent']] = df[['Price Per Unit', 'Total Spent']].astype('float')
df['Transaction Date'] = pd.to_datetime(df['Transaction Date'])
```

```python
df.isna().sum() * 100 / df.shape[0]
```

```python
df= df.dropna(subset= ['Transaction Date'])
df.info()
```

```python
df.isna().sum() * 100 / df.shape[0]
```

```python
#Filling missing columns with "INA" for Payment method and Location instead of deleting as it gives more missing info for calutions

df[['Payment Method', 'Location']] = df[['Payment Method', 'Location']].fillna('INA')
```

```python
df.isna().sum() * 100 / df.shape[0]
```

```python
#Replacting Item values with mode
df['Item'].mode(0)

df['Item'] = df['Item'].fillna(df['Item'].mode()[0])
```

```python
df.isna().sum() * 100/ df.shape[0]
```

```python
# Price per unit is a dependant item
#Groupby  each category for disticnt price
df.groupby('Item')['Price Per Unit'].value_counts()
```

```python
df.groupby('Item')['Price Per Unit'].median()
```

```python
#lambda is used here (when ever you pick an item groupby in that particular category and prepare median)

df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.median()))

df['Price Per Unit'] = df.groupby('Item')['Price Per Unit'].transform(lambda x: x.fillna(x.median()))

df.tail()
```

```python
df.isna().sum() * 100/ df.shape[0]
```

```python
#filling missing info of Total spent by the product of Price spent * quantity

df['Total Spent'] = df['Total Spent'].fillna(df['Quantity'] * df['Price Per Unit'])
```

```python
df.isna().sum() *100/df.shape[0]
```

```python
# Storing clean data
df.to_csv('/workspaces/data_cleaning_1/data/clean/Cafe Sales(Clean).csv', index = False)
```
